## Problem Statement

You are given a PySpark DataFrame named `employees` containing employee information.

### Dataset: `employees`

| Column | Data Type | Description |
|---|---|---|
| `emp_id` | Integer | Unique employee identifier |
| `name` | String | Employee name |
| `department` | String | Department in which the employee works |
| `years_experience` | Integer | Number of years of work experience |
| `salary` | Double | Annual employee salary |

## Task

Find all employees who have **10 or more years of experience**.

The result should:

1. Include only employees with `years_experience >= 10`.
2. Return only the following columns:
   - `emp_id`
   - `name`
   - `years_experience`
3. Sort the result by `years_experience` in **descending order**.

## Expected Output

| emp_id | name | years_experience |
|---:|---|---:|
| 3 | Carol White | 15 |
| 1 | Alice Smith | 12 |
| 5 | Eva Martinez | 10 |
