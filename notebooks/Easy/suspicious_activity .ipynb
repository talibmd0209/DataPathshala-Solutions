# Suspicious Login Activity

## Problem Description

You are given a Pandas DataFrame named **`login_attempts`** containing login attempts made by users.

### DataFrame: `login_attempts`

| Column | Data Type | Description |
|---|---|---|
| `attempt_id` | int | Unique identifier for each login attempt |
| `user_id` | int | Identifier of the user |
| `attempt_time` | str | Timestamp of the login attempt in `YYYY-MM-DD HH:MM:SS` format |
| `success` | int | `1` if the login was successful, `0` if the login failed |

## Task

Find all users who had **5 or more consecutive failed login attempts** (`success = 0`) at any point.

For each qualifying user, determine the **longest consecutive failure streak**.

### Important Rules

1. Login attempts must be ordered by `attempt_time` for each user.
2. A failed attempt (`success = 0`) increases the current failure streak.
3. A successful attempt (`success = 1`) breaks the failure streak and resets the count to `0`.
4. Do **not** filter out successful attempts before identifying consecutive groups. Successful attempts are required to separate different failure streaks.
5. A user qualifies only if their maximum failure streak is **5 or more**.
6. If a user has multiple failure streaks, return only their **longest** streak.
